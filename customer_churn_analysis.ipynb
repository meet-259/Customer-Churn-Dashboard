# Univariate Analysis

## Churn

```python
df['Churn'] = df['Churn'].map({'Yes':1, 'No':0})
```

```python
df['Churn'].value_counts()
```

```python
churn_rate = df['Churn'].mean()
print('Churn Rate -', churn_rate)
```

## Customer Demographics

```python
cust_info = ['gender', 'SeniorCitizen', 'Partner', 'Dependents']

for col in cust_info:
    print()
    print(f'Analysis of {col} column')
    print((df.groupby(col)
        .agg(
            Total_Customers=('customerID', 'count'), 
            Churned_Customers=('Churn', 'sum'),
            Churn_Rate=('Churn', lambda x: x.mean() * 100)
        )
    ))
```

## Tenure

```python
df['tenure'].describe()
```

```python
df.groupby('Churn')['tenure'].describe()
```

```python
bins = [0, 12, 24, 48, 72]
labels = ['0-12', '13-24', '25-48', '49-72']

df['tenure_group'] = pd.cut(
    df['tenure'],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```

```python
(df.groupby('tenure_group', observed=False)
    .agg(
        Total_Customers=('customerID', 'count'),
        Churned_Customers=('Churn', 'sum'),
        Churn_Rate=('Churn', lambda x: x.mean() * 100)
    )
)
```

## Service & Payment

```python
service_info = ['PhoneService', 'MultipleLines', 'InternetService','OnlineSecurity',
             'TechSupport', 'Contract', 'PaymentMethod']

for col in service_info:
    print()
    print(f'Analysis of {col} column')
    print((df.groupby(col)
        .agg(
            Total_Customers=('customerID', 'count'), 
            Churned_Customers=('Churn', 'sum'),
            Churn_Rate=('Churn', lambda x: x.mean() * 100)
        )
    ))
```

# Conclusion

During the analysis, I found that churn was concentrated in a specific customer segment rather than being driven by a single factor. The customers who churned most frequently were typically:

- Customers in their first 12 months of tenure
- Customers on month-to-month contracts
- Customers using electronic check as the payment method
- Customers with Fiber Optic internet
- Customers with relatively higher monthly charges
- Customers without a partner or dependents

Based on these insights, I recommended that the company focus its retention efforts on new customers during their first year, encourage migration to longer-term contracts, and proactively engage customers in this high-risk segment."
